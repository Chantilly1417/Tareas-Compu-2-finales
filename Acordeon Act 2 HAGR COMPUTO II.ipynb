```python
#Ahora seleccionaremos la columna que quieras de la tabla, para eso usaremos corchetes, usaremos de ejemplo la columna col1
df["col1"]
```

```python
#Si quieres seleccionar 2 columnas puedes hacerlo separandolas con un simple coma, no olvides los [""] extra 
df[["col1","col3"]]
```

```python
#En caso que requieras varias columnas, dales un nombre
Grupo_Columnas = ["col1","col2","col3","col5"]
df[Grupo_Columnas]
```

```python
#A continuación filtraremos datos de una columna con una condicion donde nos aviente True o False, si la columna tiene "Sur" la marcara como True, de lo contrario False
df["col6"] == "Sur"
```

```python
#Para acceder a la tabla de la columna condicionada solo se necesita:
df[df["col6"] == "Sur"]
```

```python
#Si ocupas mas de una condicion tu tranqui, solo separalas con parentesis y conectalos con operadores 
df[(df["col10"]=="Alto") & (df["col9"]!="No")]
```

```python
#Ahora ordenaremos las columnas del dataframe como queramos, de menos a mayor o viceversa 
#En este caso la ordenara de menor a mayor por default, en caso de querer lo contrario especificar 
df.sort_values("col1") 
```

```python
#En caso de solicitarlo de mayor a menor
df.sort_values("col1", ascending = False)
```

```python
#Para calcular la media se usa .mean()
df["col1"].mean()
```

```python
#maximo o minimo max o min
df["col1"].max()
```

```python
#Para el analizar variables de texto utilizamos value_counts() para contar cuántas veces aparece cada valor, unique() para obtener una lista de los valores distintos (osea que variables hay) o nunique() para saber la cantidad exacta de valores diferentes (cuantas variables hay).
ContarSubCategorias = df["col10"].value_counts()
display(ContarSubCategorias)
```
