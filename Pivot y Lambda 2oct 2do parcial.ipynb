```python
import pandas as pd
```

```python
orders = pd.read_csv("olist_orders_dataset.csv")
order_items = pd.read_csv("olist_order_items_dataset.csv")
order_reviews = pd.read_csv("olist_order_reviews_dataset.csv")
products = pd.read_csv("olist_products_dataset.csv")
customers = pd.read_csv("olist_customers_dataset.csv")
sellers = pd.read_csv("olist_sellers_dataset.csv")
geo = pd.read_csv("olist_geolocation_dataset.csv")
payments = pd.read_csv("olist_order_payments_dataset.csv")
```

```python
orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"],
    format="%Y-%m-%d %H:%M:%S",
    errors="coerce"
)

orders["order_delivered_customer_date"] = pd.to_datetime(
    orders["order_delivered_customer_date"],
    format="%Y-%m-%d %H:%M:%S",
    errors="coerce"
)

orders["order_estimated_delivery_date"] = pd.to_datetime(
    orders["order_estimated_delivery_date"],
    format="%Y-%m-%d %H:%M:%S",
    errors="coerce"
)
```

### Nivel 2 Pivot Tables con 2 Dimensiones

```python
tabla_cat_prod = pd.merge(products, order_items, on="product_id", how="left")
tabla_cat_prod = pd.merge(tabla_cat_prod, orders[["order_id", "order_status"]], on="order_id", how="left")
pivot = pd.pivot_table(tabla_cat_prod, index="product_category_name", columns="order_status", values="price", aggfunc="sum", fill_value=0)
display(pivot)
```

```python
#7 
tabla_cat_prod = pd.merge(products, order_items, on="product_id", how="left")
tabla_cat_prod = pd.merge(tabla_cat_prod, orders[["order_id", "order_status"]], on="order_id", how="left")
pivot = pd.pivot_table(tabla_cat_prod, index="product_category_name", columns="order_status", values="price", aggfunc="mean", fill_value=0)
display(pivot)
```

```python
#8
pivot = pd.pivot_table(tabla_cat_prod, index= "product_category_name", columns= "order_status", values= "freight_value", aggfunc= "mean", fill_value=0)
display(pivot)
```

```python
base = order_items.merge(products[["product_id", "product_category_name"]], on="product_id", how="left") \
.merge(orders[["order_id", "order_status"]], on="order_id", how="left")
```

```python
#9
tabla_original = base.pivot_table(index="product_category_name", columns="order_status", values="price", aggfunc="sum")
tabla_fill0 = base.pivot_table(index="product_category_name", columns="order_status", values="price", aggfunc="sum", fill_value=0)
tabla_original.isna().sum().sum()
```

```python
#10
base.pivot_table(index=["product_category_name", "order_status"], values="price", aggfunc="mean")
```

```python
#11 
base["total"] = base["price"] + base["freight_value"]
display(base["total"])
```

```python
# 12
base["categoria_precio"] = base["price"].apply(lambda x: 'caro' if x > 100 else 'barato')
display(base["categoria_precio"])
```

```python
# 13
base["flete_alto"] = base.apply(lambda row: row["freight_value"] > 0.2 * row["price"], axis=1)
display(base["flete_alto"])
```

```python
# 14
base["categoria_mayus"] = base["product_category_name"].apply(lambda x: x.upper() if pd.notna(x) else x)
display(base["categoria_mayus"])
```

```python
# 15
base["contiene_palabra"] = base["product_category_name"].apply(lambda x: 'beleza' in x if pd.notna(x) else False)
display(base["contiene_palabra"])
```

```python
# 16
base["segmento_costo"] = base["total"].apply(lambda x: 'alto' if x > 150 else 'bajo')
base.pivot_table(index="product_category_name", columns="segmento_costo", values="total", aggfunc="mean")
```

```python
#17
base["pct_envio"] = base["freight_value"] / base["price"]
base["clasificacion_envio"] = base["pct_envio"].apply(lambda x: 'envío alto' if x > 0.2 else 'envío bajo')
base.pivot_table(index="clasificacion_envio", values="price", aggfunc=["mean", "count"])
```

```python
#18
precio_promedio_global = base["price"].mean()
base["sobre_promedio"] = base["price"].apply(lambda x: x > precio_promedio_global)
base.pivot_table(index="sobre_promedio", values="price", aggfunc=["mean", "count"])
```

```python
#19
base.pivot_table(index="product_category_name", values="freight_value", aggfunc=["sum", "mean"])
```

```python
#20
base["categoria_precio"] = base["price"].apply(lambda x: 'caro' if x > 100 else 'barato')
base["pct_envio"] = base["freight_value"] / base["price"]
tabla_20 = base.pivot_table(index="product_category_name", columns="categoria_precio", values="pct_envio", aggfunc="mean")
display(tabla_20)
```

```python
#21
base["ratio_envio_precio"] = base.apply(lambda row: row["freight_value"] / row["price"] if row["price"] > 0 else None, axis=1)
base.pivot_table(index="product_category_name", values="ratio_envio_precio", aggfunc="mean").sort_values("ratio_envio_precio", ascending=False)
```

```python
#22
tabla_precio_estado = base.pivot_table(index="product_category_name", columns="order_status", values="price", aggfunc="mean")
base["flete_alto"] = base.apply(lambda row: row["freight_value"] > 0.2 * row["price"], axis=1)
tabla_flete_estado = base.pivot_table(index="product_category_name", columns="order_status", values="flete_alto", aggfunc="mean")
display(tabla_flete_estado)
```

```python
#Reto 
base["segmento_precio"] = base["price"].apply(lambda x: "caro" if x > 100 else "barato")
base["ratio_envio_precio"] = base.apply(
    lambda row: row["freight_value"] / row["price"] if row["price"] > 0 else None,
    axis=1)
resumen_direccion = base.pivot_table(index=["product_category_name", "segmento_precio"], columns="order_status", values=["price", "freight_value", "ratio_envio_precio"], aggfunc={"price": "mean", "freight_value": "mean", "ratio_envio_precio": "mean"}, fill_value=0)
display(resumen_direccion)
```
